```python
import numpy as np
import pandas as pd


def generate_logistics_dataset():
  np.random.seed(202)
  n_rows = 1200

  providers = [
      "JNE Express",
      "J&T Express",
      "SiCepat Ekspres",
      "AnterAja",
      "GoSend",
      "GrabExpress",
  ]
  provider_weights = [0.25, 0.25, 0.20, 0.15, 0.10, 0.05]

  delivery_types = ["Regular", "Express", "Same-Day", "Cargo"]
  type_weights = [0.50, 0.25, 0.15, 0.10]

  provinces = [
      "DKI Jakarta",
      "Jawa Barat",
      "Jawa Timur",
      "Jawa Tengah",
      "Banten",
      "Sumatera Utara",
      "Bali",
  ]

  data = []
  for i in range(n_rows):
    provider = np.random.choice(providers, p=provider_weights)
    dtype = np.random.choice(delivery_types, p=type_weights)
    province = np.random.choice(provinces)

    # Estimasi dan biaya berdasarkan jenis pengiriman
    if dtype == "Same-Day":
      est_days = 0
      cost = int(np.random.randint(25000, 60000))
    elif dtype == "Express":
      est_days = 1
      cost = int(np.random.randint(18000, 35000))
    elif dtype == "Regular":
      est_days = int(np.random.choice([2, 3, 4]))
      cost = int(np.random.randint(9000, 20000))
    else:  # Cargo
      est_days = int(np.random.choice([5, 6, 7]))
      cost = int(np.random.randint(50000, 150000))

    # Simulasi keterlambatan pengiriman
    delay_chance = 0.15 if dtype != "Same-Day" else 0.05
    is_delayed = np.random.choice(
        [True, False], p=[delay_chance, 1 - delay_chance]
    )

    if is_delayed:
      actual_days = est_days + int(np.random.choice([1, 2, 3]))
      status = "Delayed"
      rating = round(
          float(np.random.choice([2.5, 3.0, 3.5, 4.0], p=[0.3, 0.4, 0.2, 0.1])),
          1,
      )
    else:
      actual_days = est_days
      status = "On-Time"
      rating = round(
          float(np.random.choice([4.0, 4.5, 4.8, 5.0], p=[0.1, 0.2, 0.3, 0.4])),
          1,
      )

    data.append({
        "order_id": f"LOG-{7000 + i}",
        "shipping_provider": provider,
        "delivery_type": dtype,
        "destination_province": province,
        "shipping_cost_idr": cost,
        "estimated_days": est_days,
        "actual_days": actual_days,
        "delivery_status": status,
        "customer_satisfaction_rating": rating,
    })

  df = pd.DataFrame(data)
  filename = "indonesia_ecommerce_logistics.csv"
  df.to_csv(filename, index=False)
  print(
      f"Success! Dataset generated with {len(df)} records and saved as"
      f" '{filename}'"
  )


if __name__ == "__main__":
  generate_logistics_dataset()
```
